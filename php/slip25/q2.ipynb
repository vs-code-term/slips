```python
import nltk
nltk.download('punkt')


import pandas as pd
from nltk.tokenize import word_tokenize
from textblob import TextBlob


df = pd.read_csv("covid_2021_1.csv")

# Display first rows
print(df.head())



df = df.drop_duplicates()


df = df.dropna()


comments = df['comment_text']






print("\nTokenized comments:")
for i in comments.head(5):
    tokens = word_tokenize(i)
    print(tokens)



positive = 0
negative = 0
neutral = 0

for comment in comments:
    analysis = TextBlob(comment)
    polarity = analysis.sentiment.polarity

    if polarity > 0:
        positive += 1
    elif polarity < 0:
        negative += 1
    else:
        neutral += 1

total = len(comments)


pos_percent = (positive / total) * 100
neg_percent = (negative / total) * 100
neu_percent = (neutral / total) * 100


print("\nSentiment Analysis Results:")
print("Positive: {:.2f}%".format(pos_percent))
print("Negative: {:.2f}%".format(neg_percent))
print("Neutral: {:.2f}%".format(neu_percent))
```
