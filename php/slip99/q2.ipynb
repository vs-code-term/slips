```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
from sklearn.metrics import r2_score, mean_squared_error
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression

data = pd.read_csv("D:/datasets/Salary.csv")

Y = data['Salary']
X = data.drop(['Salary'], axis=1)

x_train, x_test, y_train, y_test = train_test_split(X, Y, test_size=0.25, random_state=0)

slr = LinearRegression()
slr.fit(x_train, y_train)

y_pred = slr.predict(x_test)
x_pred = slr.predict(x_train)

print(f"Training Data (X):\n{x_train}\n")
print(f"Testing Data (X):\n{x_test}\n")
print(f"Predictions:\n{y_pred}\n")
print(f"Actual vs Predicted:\n{pd.DataFrame({'Actual': y_test, 'Predicted': y_pred})}\n")
print(f"Prediction for 3 years of experience: {slr.predict([[3]])[0]} ₹")

plt.scatter(x_train, y_train, color="green")
plt.plot(x_train, x_pred, color="red")
plt.title("Salary vs Experience (Training Dataset)")
plt.xlabel("Years of Experience")
plt.ylabel("Salary(In Rupees)")
plt.show()

print(f'R2 score: {r2_score(y_test, y_pred):.2f}')
print(f'Mean Squared Error: {mean_squared_error(y_test, y_pred)}')

def salary(exp):
    return slr.predict([[exp]])[0]

exp = float(input('Enter experience in years: '))
print(f'Salary will be: {salary(exp)} ₹')
```
