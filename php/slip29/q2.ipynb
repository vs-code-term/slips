```python
from sklearn.linear_model import LogisticRegression


hours = [[1], [2], [3], [4], [5], [6], [7], [8]]
result = [0, 0, 0, 0, 1, 1, 1, 1]

model = LogisticRegression()


model.fit(hours, result)


pred = model.predict([[3]])
print("Prediction for 3 hours (0=Fail,1=Pass):", pred[0])

pred2 = model.predict([[7]])
print("Prediction for 7 hours (0=Fail,1=Pass):", pred2[0])


prob = model.predict_proba([[5]])
print("Probability for 5 hours [Fail, Pass]:", prob[0])
```
