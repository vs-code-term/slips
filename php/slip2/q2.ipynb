```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
from sklearn.metrics import r2_score, mean_squared_error
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression

data = pd.read_csv("D:/datasets/Salary.csv")

print(data)
print(data.shape)
print(data.describe())
print(data.isnull())
print(data.head(10))

corr = data.corr()
print(corr)

print(data.dtypes)

Y = data['Salary']
X = data.drop(['Salary'], axis=1)

print(Y)
print(X)

plt.scatter(X, Y)
plt.title("Salary vs Experience (Training Dataset)")
plt.xlabel("Experience in Year")
plt.ylabel("Salary(In Rupees)")
plt.show()

x_train, x_test, y_train, y_test = train_test_split(X, Y, test_size=0.25, random_state=0)

print(x_train)
print(y_train)
print(x_test.shape)

slr = LinearRegression()
slr.fit(x_train, y_train)

y_pred = slr.predict(x_test)
x_pred = slr.predict(x_train)

print(y_pred)

df = pd.DataFrame({'Actual': y_test, 'predicted': y_pred})
print(df)

print(slr.predict([[3]]))

plt.scatter(x_train, y_train, color="green")
plt.plot(x_train, x_pred, color="red")
plt.title("Salary vs Experience (Training Dataset)")
plt.xlabel("Years of Experience")
plt.ylabel("Salary(In Rupees)")
plt.show()

print('R2 score: %.2f' % r2_score(y_test, y_pred))
print('Mean Error :', mean_squared_error(y_test, y_pred))

def salary(exp):
    result = slr.predict([[exp]])
    return result

exp = float(input('Enter experience in years : '))
print('Salary will be : ', float(salary(exp)), '₹')
```
